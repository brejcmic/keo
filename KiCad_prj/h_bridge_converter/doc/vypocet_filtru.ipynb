```python
# Parametry výkonového systému
class PowerSystem:
    def __init__(self):
        
        # Frekvence PWM
        self.freq_pwm = 30000;
        
        # Vstupní napětí
        self.voltin_ac = 230;
        self.voltin_dc = 230 * (2**0.5);
```

```python
# Tvorba objektu výkonového systému
power_system = PowerSystem();
```

```python
# Parametry pro výpočet výstupního obvodu filtru
class FilterParams:
    def __init__(self):
        
        # Zvlnění proudu cívkou
        self.curr_delta = [0.5 * i for i in range(1,5)];
        
        # Zvlnění výstupního napětí
        self.volt_delta = 1;
        
        # Indukčnost v závislosti na zvlnění proudu
        self.inductance = [0 for i in range(1,5)];
        
        # Kapacita v závislosti na zvlnění proudu
        self.capacitance = self.inductance;
        
        # Vlastní frekvence uskupení
        self.resonance = self.inductance;
        
    def evaluate(self, ps:PowerSystem):
        
        # zde je výpočet celkové indukčnosti, ta je sestavena ze dvou cívek pro bipolární řízení, 
        # při unipolárním řízení je výsledná índukčnost 1/4
        self.inductance = [ps.voltin_dc / ps.freq_pwm / 2 / current_delta for current_delta in self.curr_delta];
        
        # výpočet kapacity - předpoklad bipolárního řízení, tedy čas je polovinou frekvence PWM
        self.capacitance = [current_delta / ps.freq_pwm / 2 / self.volt_delta for current_delta in self.curr_delta];
        
        # vlastní frekvence uskupení
        self.resonance = [1 / 2 / 3.1415 / ((ind * cap)**0.5) for ind, cap in zip(self.inductance, self.capacitance)];
        
        print(f"Indukčnost a kapacita v závislosti na velikosti změny proudu:\n");
        
        for i in range(len(self.inductance)):
            print(f"{self.curr_delta[i]} A, {self.inductance[i]} H, {self.capacitance[i]} F, {self.resonance[i]} Hz");
            
```

```python
# Výpočet filtru
filter_params = FilterParams();
filter_params.evaluate(power_system);
```
